```python
import websocket
import ssl

IP = "11.0.10.2"
PORT = 8090

paths = [
    "/ws",
    "/ws/",
    "/websocket",
    "/websocket/",
    "/asterisk/ws",
    "/ari",
    "/ari/events",
]

for path in paths:
    url = f"wss://{IP}:{PORT}{path}"

    print("\n" + "=" * 70)
    print(f"Probando: {url}")

    try:
        ws = websocket.create_connection(
            url,
            timeout=5,
            sslopt={
                "cert_reqs": ssl.CERT_NONE
            }
        )

        print("✅ ¡CONECTADO!")
        print(f"Endpoint válido: {url}")

        ws.close()

    except websocket.WebSocketBadStatusException as e:
        print(f"❌ HTTP {e.status_code}")
        print(str(e))

    except Exception as e:
        print(f"❌ {type(e).__name__}: {e}")
```

```python
import ssl
import websocket

IP = "11.0.10.2"   # <-- cambia por la IP de tu UCM6308
PORT = 8090

url = f"wss://{IP}:{PORT}/ws"

print(f"Probando: {url}")

try:
    ws = websocket.create_connection(
        url,
        timeout=5,
        sslopt={
            "cert_reqs": ssl.CERT_NONE
        }
    )

    print("✅ WSS CONECTADO")
    print("El puerto acepta una conexión WebSocket segura.")

    ws.close()

except websocket.WebSocketBadStatusException as e:
    print("⚠️ El servidor respondió HTTP pero rechazó el WebSocket.")
    print(f"Status: {e.status_code}")
    print(f"Mensaje: {e.msg}")

except ssl.SSLError as e:
    print("❌ Error TLS/SSL")
    print(e)

except ConnectionRefusedError:
    print("❌ Conexión rechazada. El puerto está cerrado o el servicio no escucha.")

except TimeoutError:
    print("❌ Timeout. No se pudo establecer la conexión.")

except Exception as e:
    print("❌ Error:")
    print(type(e).__name__, e)
```
